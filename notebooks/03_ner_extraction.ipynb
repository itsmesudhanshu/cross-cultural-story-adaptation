```python
import spacy

nlp = spacy.load("en_core_web_sm")

text = "Tom celebrated Christmas in New York."

doc = nlp(text)

for ent in doc.ents:
    print(ent.text, "->", ent.label_)
```

```python
def extract_entities(text):

    doc = nlp(text)

    entities = []

    for ent in doc.ents:
        entities.append({
            "entity": ent.text,
            "label": ent.label_
        })

    return entities
```

```python
extract_entities(
    "Tom celebrated Christmas in New York."
)
```
