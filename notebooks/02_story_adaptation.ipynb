```python
def adapt_story(story):

    prompt = f"""
    Convert the following Western story into
    an Indian cultural setting.

    Rules:
    1. Preserve moral lesson.
    2. Preserve plot.
    3. Replace names.
    4. Replace festivals.
    5. Replace food and locations.

    Story:
    {story}
    """

    response = model.generate_content(prompt)

    return response.text
```

 ### Cultural Mapping Dictionary



```python
cultural_mapping = {
    
    # Names
    "Tom": "Rohan",
    "Jack": "Arjun",
    "Sujan": "Sujata",
    "Bryan": "Brijesh",
    "Maddy": "Madhav",

    # Festivals
    "Christmas": "Diwali",
    "Thanksgiving": "Pongal",
    "Halloween": "Holi",

    # Places
    "New York": "Delhi",
    "London": "Mumbai",
    "California": "Bengaluru",

    # Buildings
    "castle": "Raj Mahal",
    "palace": "Mahal",

    # People
    "king": "Maharaja",
    "queen": "Maharani",
    "knight": "warrior",

    # Food
    "apple pie": "gulab jamun",
    "burger": "vada pav",
    "cookies": "laddoo",

    # Celebrations
    "Christmas tree": "diyas and rangoli",
    "Santa Claus": "grandparents"
}
```

```python
def basic_adapt(story):

    adapted_story = story

    for old, new in cultural_mapping.items():
        adapted_story = adapted_story.replace(old, new)

    return adapted_story
```

```python
story = """
Sujan and Bryan on last year Thanksgiving donated 25 percent of salary to the orphanage.
Maddy visited his home to celebrate Christmas with his family so that he can spend quality time with family.

"""

print(basic_adapt(story))


```
